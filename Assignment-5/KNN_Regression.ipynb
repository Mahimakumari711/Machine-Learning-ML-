```python
#Defining Euclidean distance functions
def euclidean_distance(p1,p2):
    return math.sqrt(sum((a - b) ** 2 for a,b in zip(p1,p2)))
```

```python
#Define KNN Regression function
def knn_classify(train_data, test_point, k):
    distances = []
    for i in range(len(train_data)):
        dist = euclidean_distance(test_point, train_data[i])
        distances.append((dist, y_train[i]))
    print("Distances: \n",distances)
    
    #Sort distances in non decreasing order
    distances.sort(key = lambda x: x[0])
    k_nearest = [label for (_, label) in distances[: k]]
    
    #Mean Value
    y_hat = (sum(k_nearest))/k
    return y_hat
```
