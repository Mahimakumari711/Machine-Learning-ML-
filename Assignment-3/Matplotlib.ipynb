## Our first Visualization

```python
# Matplot is the library used for creating visualizations 
# pyplot is the module responsible for creating and displaying plots/graphs. 
```

```python
#Importing pyplot
#from matplotlib import pyplot as plt
import matplotlib.pyplot as plt

#plotting to our canvas 
plt.plot([1,2,3],[4,5,1])

#Showing what we plotted
plt.show()
```

```python
x=[1,2,3]
y=[4,5,1]
plt.plot(x,y)
plt.show()
```

# Adding title and labelling

```python
import matplotlib.pyplot as plt

x=[5,8,10]
y=[12,16,6]

plt.plot(x,y)

plt.title("Plotting information")

#putting label to both the axis
plt.ylabel('y axis')
plt.xlabel('x axis')

plt.show()
```

# Adding style 

```python
import matplotlib.pyplot as plt
import matplotlib.style 

x=[5,8,10]
y=[12,16,6]

plt.plot(x,y,color='red' , linewidth= 5 )

plt.ylabel('y axis')
plt.xlabel('x axis')

plt.show()
```
